```python
import requests

gene_id = "ENSRNOG00000062930"
server = "https://rest.ensembl.org"
ext = f"/homology/id/{gene_id}?content-type=application/json"

r = requests.get(server + ext, headers={"Content-Type": "application/json"})
if not r.ok:
    r.raise_for_status()

data = r.json()
for homology in data['data'][0]['homologies']:
    print(f"{homology['target']['species']}: {homology['target']['id']}")
```
