```python
import pandas as pd
```

```python
degs_24W_M = pd.read_csv('/mnt/scratch/bowmand8/scVIDR/scvidr_sandbox/scVIDR/data/DEGs_Brown_Adipocytes/taPVAT_Adipocytes_Brown_HF_24W_M_vs_Control_24W_M_diet_comparison_deseq_all_genes.txt', sep='\t')
```

```python
degs_24W_M
```

```python
predictions_24W_M = pd.read_csv('/mnt/scratch/bowmand8/scVIDR/scvidr_sandbox/scVIDR/predictions/gene_scores/August18_10_18_gene_scores_Adipocytes_Brown_24W_M.csv')
```

```python
predictions_24W_M = predictions_24W_M[~predictions_24W_M['Gene'].str.startswith('Mt-')] # Removed mitochondrial genes
predictions_24W_M = predictions_24W_M[~predictions_24W_M['Gene'].str.startswith('AABR')]
predictions_24W_M = predictions_24W_M[~predictions_24W_M['Gene'].str.startswith('ENSRN')]
```

```python
degs_24W_M = degs_24W_M[degs_24W_M['padj'] <= 0.05]
degs_24W_M = degs_24W_M[~degs_24W_M['gene_id'].str.startswith('Mt-')] # Removed mitochondrial genes
degs_24W_M = degs_24W_M[~degs_24W_M['gene_id'].str.startswith('AABR')]
degs_24W_M = degs_24W_M[~degs_24W_M['gene_id'].str.startswith('ENSRN')]

degs_24W_M
```

```python
degs_up_in_HF = degs_24W_M[degs_24W_M['log2FoldChange'] >= 1.0]
degs_down_in_HF = degs_24W_M[degs_24W_M['log2FoldChange'] <= -1.0]
```

```python
degs_down_in_HF
```

```python
upregulated_predictions_24W_M = predictions_24W_M[predictions_24W_M['Gene'].isin(list(degs_up_in_HF['gene_id']))]
downregulated_predictions_24W_M = predictions_24W_M[predictions_24W_M['Gene'].isin(list(degs_down_in_HF['gene_id']))]
```

```python
gene_to_logfc = degs_24W_M.set_index('gene_id')['log2FoldChange']
downregulated_predictions_24W_M['GroundTruthLog2FC'] = downregulated_predictions_24W_M['Gene'].map(gene_to_logfc)
upregulated_predictions_24W_M['GroundTruthLog2FC'] = upregulated_predictions_24W_M['Gene'].map(gene_to_logfc)
```

```python
upregulated_predictions_24W_M.sort_values(by='Score', inplace=True, ascending=False)
downregulated_predictions_24W_M.sort_values(by='Score', inplace=True, ascending=True)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.linear_model import LinearRegression


# Top 25 most positive gene scores
top_pos = upregulated_predictions_24W_M.head(25)

plt.figure(dpi=300)
plt.bar(top_pos['Gene'], top_pos['Score'], color='darkgreen')
plt.scatter(top_pos['Gene'], top_pos['GroundTruthLog2FC'], color='black', label='Ground Truth Log2FC')



X = np.arange(len(top_pos['Gene'])).reshape(-1, 1)
y = top_pos['GroundTruthLog2FC'].values


ols = LinearRegression()
ols.fit(X, y)

y_pred = ols.predict(X)






plt.plot(top_pos['Gene'], y_pred, color='blue', linestyle='--', label='OLS Fit (MSE)')
plt.legend()

plt.xticks(rotation=90, fontsize=12)
plt.ylabel("scVIDR Gene Score", fontsize=14)
plt.title("Top 25 Positive Gene Scores", fontsize=16)
plt.tight_layout()
#plt.savefig("../figures/Top25PositiveGeneScores_24W_Brown_Adipocytes_Both.svg", bbox_inches="tight")
plt.show()




# Top 25 most negative gene scores
top_neg = downregulated_predictions_24W_M.head(25)
plt.figure(dpi=300)
plt.bar(top_neg['Gene'], top_neg['Score'], color='darkred')
plt.scatter(top_neg['Gene'], top_neg['GroundTruthLog2FC'], color='black', label='Ground Truth Log2FC')


X = np.arange(len(top_neg['Gene'])).reshape(-1, 1)
y = top_neg['GroundTruthLog2FC'].values


ols = LinearRegression()
ols.fit(X, y)

y_pred = ols.predict(X)

plt.plot(top_neg['Gene'], y_pred, color='blue', linestyle='--', label='OLS Fit (MSE)')
    
plt.xticks(rotation=90, fontsize=12)
plt.ylabel("scVIDR Gene Score", fontsize=14)
plt.title("Top 25 Negative Gene Scores", fontsize=16)
plt.legend()
plt.tight_layout()
#plt.savefig("../figures/Top25NegativeGeneScores_24W_Brown_Adipocytes_Both.svg", bbox_inches="tight")
plt.show()
```
